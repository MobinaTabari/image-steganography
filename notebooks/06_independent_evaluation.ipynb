# Independent Model Evaluation

## 1. Unseen UCID Image Selection

This notebook prepares an independent evaluation dataset
using UCID images excluded from previous experiments.

The selected models will be evaluated without further
hyperparameter tuning or feature selection.

The goal is to estimate generalization performance
on previously unseen source images.

```python
from pathlib import Path
import csv

project_root = Path.cwd().resolve()

if project_root.name == "notebooks":
    project_root = project_root.parent

ucid_dir = project_root / "dataset" / "ucid_raw"
metadata_path = project_root / "dataset" / "ucid_metadata.csv"

# Inspect available UCID images
ucid_images = sorted(ucid_dir.rglob("*.tif"))

with open(metadata_path, "r", encoding="utf-8", newline="") as file:
    reader = csv.DictReader(file)
    metadata = list(reader)
    columns = reader.fieldnames

print("Total UCID images available:", len(ucid_images))
print("Previously selected sources:", len(metadata))
print("Metadata columns:", columns)
print("First UCID files:", [p.name for p in ucid_images[:5]])

if metadata:
    print("Metadata example:", metadata[0])
```
